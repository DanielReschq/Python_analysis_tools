```python
dict_index = 1

eta = 0.87
nu = 1.0
gc = 1.3425

c=0#0.847454752328233
omega=0#-0.30240862978142213

k = 0.33
d = 0.01

dataM = FS3.load(parameter_dictT["path_to_data"] + 'mag2%s.dat'%(parameter_dictT["sigma"][dict_index]), LIndex=0, xIndex=1, yIndex=2, dyIndex=3)
dataU = FS3.load(parameter_dictT["path_to_data"] + 'Binder%s.dat'%(parameter_dictT["sigma"][dict_index]), LIndex=0, xIndex=1, yIndex=2, dyIndex=3)
dataM = FS3.select(dataM, gRange=parameter_dictT["gRange"][dict_index])
dataU = FS3.select(dataU, gRange=parameter_dictT["gRange"][dict_index])

Llist = np.sort(list(dataM), axis=0)
data = {}
for L in Llist:
    Tlist = dataU[L][:, 0]
    data[L] = np.empty((0, 4))
    for T in Tlist:
        iU = np.argwhere(dataU[L][:, 0] == T)[0, 0]
        iM = np.argwhere(dataM[L][:, 0] == T)[0, 0]
        data[L] = np.vstack(
            (
                data[L],
                np.array(
                    [
                        dataU[L][iU, 1],
                        dataU[L][iU, 2],
                        dataM[L][iM, 1],
                        dataM[L][iM, 2],
                    ]
                ),
            )
        )

FS3.setStyle(Llist=np.sort(list(dataM),axis=0))
fig, ax0 = FS3.figure(xlabel=r'$T$', ylabel=r'$m^2$')
for L in np.sort(list(dataM.keys()), axis=0):
    ax0.errorbar(dataM[L][:,0], dataM[L][:,1]*L**(eta),  yerr=dataM[L][:,2], **FS3.errorbarStyle(L), linestyle='', label=r'$L=%d$'%L)
ax0.legend(loc=(0.71,0.72))

FS3.setStyle(Llist=np.sort(list(data),axis=0))
fig, ax0 = FS3.figure(xlabel=r'$U_2$', ylabel=r'$m^2$')
for L in np.sort(list(data.keys()), axis=0):
    ax0.errorbar(data[L][:,0], data[L][:,2], xerr=data[L][:,1], yerr=data[L][:,3], **FS3.errorbarStyle(L), linestyle='', label=r'$L=%d$'%L)
    xrange = np.linspace(data[L][-1,0],data[L][0,0],100)
    y = L**(-eta)*(k*xrange +d)
    ax0.plot(xrange,y)
ax0.legend(loc=(0.71,0.72))
#ax0.set_xlim(dataU[128][-1, 1], dataU[128][0, 1])
#ax0.set_ylim(dataM[128][-1, 1], dataM[128][0, 1])

fig, ax1 = FS3.addInset(fig, loc=3)
# dataCollapseM = FS3.rescaleAxis(data,
#                                 xfunc="x",
#                                 yfunc="y*L**(-eta)",
#                                 arg={"eta": eta, "c": c, "omega": omega},
#                                 )
for L in np.sort(list(data), axis=0):
    ax1.errorbar(
        data[L][:, 0],
        data[L][:, 2]*L**eta,
        yerr=data[L][:, 3]*L**eta,
        **FS3.errorbarStyle(L),
        linestyle=""
    )
    #ax1.plot(xrange,k*xrange+d)
```
